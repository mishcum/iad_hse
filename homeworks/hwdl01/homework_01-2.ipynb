В данном задании нет цели побить какой-то порог по метрике. Ваша задача &mdash; убедиться в том, что ваш рукописный цикл обучения работает. Для ускорения вычислений и обучения модели можете брать только срез данных, а не весь датасет.

```python
def batch_loss(model, X, y, batch_size=1024):
    total_loss, total_n = 0.0, 0
    for i in range(0, len(X), batch_size):
        xb = torch.tensor(X[i:i+batch_size], dtype=torch.float32)
        yb = y[i:i+batch_size]
        preds = model(xb).detach().cpu().numpy()
        total_loss += np.sum((yb - preds) ** 2)
        total_n += len(xb)
    return (total_loss / total_n) ** 0.5

def loss(preds : np.ndarray, true : np.ndarray):
    return np.mean((true - preds) ** 2)

def grad(X: np.ndarray, preds: np.ndarray, true: np.ndarray):
    N = X.shape[0]
    error = preds - true
    dW = (2 / N) * X.T.dot(error)
    db = (2 / N) * np.sum(error)
    return dW, db

lr = 0.01
model_linear = nn.Linear(in_features=90, out_features=1)

print(f'Initial loss: {batch_loss(model_linear, X_test, y_test)}')

for i in range(1000):
    with torch.no_grad():
        ind = np.random.choice(len(X_train), size=10000)

        X_batch = X_train[ind]
        y_batch = y_train[ind].reshape(-1, 1)

        preds = model_linear(torch.tensor(X_batch, dtype=torch.float32)).cpu().numpy()

        dW, dB = grad(X_batch, preds, y_batch)

        model_linear.weight.data -= lr * torch.tensor(dW.T, dtype=torch.float32)
        model_linear.bias.data -= lr * torch.tensor(dB, dtype=torch.float32)

print(f'Final loss: {batch_loss(model_linear, X_test, y_test)}')
```

## Часть II. Заводим нейронную сеть (максимум 10 баллов)

Ниже нам предстоит реализовать довольно много различных нейросетей и поставить целую серию экспериментов. Чтобы это всё происходило без боли и страданий, нам нужно держать код в удобном виде.

При решении заданий вы можете придерживаться любой адекватной струкуры кода, но мы советуем воспользоваться сигнатурами функций, которые приведены ниже. При необходимости вы можете добавить в них любые нужные вам аргументы и любой нужный функционал. Более того, хорошей практикой является не делать эти функции слишком громоздкими и выносить разные хитрые штуки в отдельные функции.

```python
def training_epoch(model, optimizer, criterion, train_loader, e, epochs, device):
    """Одна эпоха обучения
    params:
        model - torch.nn.Module to be fitted
        optimizer - model optimizer
        criterion - loss function from torch.nn
        train_loader - torch.utils.data.Dataloader with train set
    """
    sum_loss = 0
    cnt = 0
    tqdm_loader = tqdm(train_loader)
    model.train()
    for x_train, y_train in tqdm_loader:
        x_train = x_train.to(device)
        y_train = y_train.to(device)

        preds = model(x_train)

        loss = criterion(preds.squeeze(), y_train.float())

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        sum_loss += loss.item()
        cnt += 1
        tqdm_loader.set_description(f"Epoch [{e + 1}/{epochs}], loss_mean={(sum_loss/cnt):.3f}")

    return sum_loss / cnt

@torch.no_grad()
def validation_epoch(model, criterion, val_loader, device):
    """Одна эпоха валидации модели
    params:
        model - torch.nn.Module to be fitted
        criterion - loss function from torch.nn
        val_loader - torch.utils.data.Dataloader with test set
                      (if you wish to validate during training)
    """
    model.eval()
    sum_loss, cnt = 0, 0
    for x_val, y_val in val_loader:
        x_val = x_val.to(device)
        y_val = y_val.to(device)

        val_preds = model(x_val)
        loss = criterion(val_preds.squeeze(), y_val).item()
        sum_loss += loss
        cnt += 1

    return sum_loss / cnt

@torch.no_grad()
def validation_epoch(model, criterion, val_loader, device):
    model.eval()
    sum_loss, cnt = 0, 0
    all_preds, all_targets = [], []
    for x_val, y_val in val_loader:
        x_val = x_val.to(device)
        y_val = y_val.to(device)
        val_preds = model(x_val).squeeze()
        loss = criterion(val_preds, y_val.float()).item()
        sum_loss += loss
        cnt += 1
        all_preds.append(val_preds.cpu())
        all_targets.append(y_val.cpu())

    all_preds = torch.cat(all_preds)
    all_targets = torch.cat(all_targets)

    bias2 = ((all_preds.mean() - all_targets.mean())**2).item()
    variance = torch.var(all_preds).item()

    return sum_loss / cnt, bias2, variance


def training_plot(train_losses, val_losses, biases, variances, e):
    clear_output(True)
    fig, ax = plt.subplots(figsize=(10, 5), ncols=2)

    ax[0].plot(train_losses, label='Train loss')
    ax[0].plot(val_losses, label='Validation loss')
    ax[0].set_xlabel("Epoch")
    ax[0].set_ylabel("Loss")
    ax[0].legend(loc="upper left")

    ax[1].plot(biases, 'r--', label='Bias²')
    ax[1].plot(variances, 'g--', label='Variance')
    ax[1].set_ylabel("Bias–Variance")
    ax[1].legend(loc="upper right")

    plt.suptitle(f"Training + Bias–Variance, epoch {e+1}")
    plt.show()

@torch.no_grad()
def predict(model, data_loader):
    """ Предсказания модели
    params:
        model - torch.nn.Module to be evaluated on test set
        criterion - loss function from torch.nn
        data_loader - torch.utils.data.Dataloader with test set
    ----------
    returns:
        predicts - torch.tensor with shape (len(test_loader.dataset), ),
                   which contains predictions for test objects
    """
    model.eval()
    preds_list = []

    for X_batch, _ in data_loader:
        preds = model(X_batch)

        preds_list.append(preds.cpu())

    predicts = torch.cat(preds_list, dim=0)
    return predicts

def train(model, optimizer, criterion, train_loader, val_loader, epochs, device):
    """ Обучение модели
    params:
        model - torch.nn.Module to be fitted
        optimizer - model optimizer
        criterion - loss function from torch.nn
        train_loader - torch.utils.data.Dataloader with train set
        val_loader - torch.utils.data.Dataloader with test set
                      (if you wish to validate during training)
        epochs - number of training epochs
    """
    train_losses = []
    val_losses = []
    biases = []
    variances = []
    for e in range(epochs):
        train_loss = training_epoch(model, optimizer, criterion, train_loader, e, epochs, device)

        val_loss, bias, variance = validation_epoch(model, criterion, val_loader, device)

        if e > 0:
            train_losses.append(train_loss)
            val_losses.append(val_loss)
            biases.append(bias)
            variances.append(variance)

            training_plot(train_losses, val_losses, biases, variances, e)
        print(f'Epoch [{e+1}/{epochs}] | loss: {train_loss:.3f} | val_loss: {val_loss:.3f}')
        torch.save(model.state_dict(), f=f'models/epoch_{e+1}.tar')

    return model.eval()
```

```python
classes = {c : i for i, c in enumerate(np.unique_values(y))}
classes_to_labels = {i : c for i, c in enumerate(np.unique_values(y))}
```

```python
def code_target(y, classes):
    return np.array([classes[int(c)] for c in y])
```

```python
train_dataset = torch.utils.data.TensorDataset(torch.tensor(X_train, dtype=torch.float32), torch.tensor(code_target(y_train, classes), dtype=torch.float32))
val_dataset = torch.utils.data.TensorDataset(torch.tensor(X_val, dtype=torch.float32), torch.tensor(code_target(y_val, classes), dtype=torch.float32))
test_dataset = torch.utils.data.TensorDataset(torch.tensor(X_test, dtype=torch.float32), torch.tensor(code_target(y_test, classes), dtype=torch.float32))
```

```python
device = torch.device('mps')

model = nn.Sequential(
    nn.Linear(in_features=90, out_features=128),
    nn.ReLU(),
    nn.Linear(128, 89)
)

model.to(device)

optimizer = torch.optim.SGD(params=model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()

train_loader = torch.utils.data.DataLoader(dataset=train_dataset, batch_size=64)
test_loader = torch.utils.data.DataLoader(dataset=test_dataset, batch_size=64)
val_loader = torch.utils.data.DataLoader(dataset=val_dataset, batch_size=64)

epochs = 100
```

```python
train(model, optimizer, criterion, train_loader, val_loader, epochs, device)
```

```python
def decode_target(y, classes_to_labels):
    return np.array([classes_to_labels[int(c)] for c in y])
```

```python
model.to('cpu')
preds = predict(model, test_loader)
preds = torch.softmax(preds, dim=1)
preds = torch.argmax(preds, dim=1)
preds = preds.numpy()
preds = decode_target(preds, classes_to_labels)
```

```python
from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(y_test, preds))
```

Реализуйте функции `normalize` и `denormalize`, которые, соответственно, нормируют целевую переменную и применяют обратное преобразование.
```python
class ResidualBlock(nn.Module):
    def __init__(self, dim, hidden_dim, p=0.2):
        super().__init__()
        self.block = nn.Sequential(
            nn.Linear(dim, hidden_dim),
            nn.GELU(),
            nn.Dropout(p),
            nn.Linear(hidden_dim, dim, bias=False),
            nn.BatchNorm1d(dim)
        )
        self.activation = nn.ReLU()

    def forward(self, x):
        out = self.block(x)
        out = out + x
        return self.activation(out)


class TabularNet(nn.Module):
    def __init__(self, in_features=90, hidden_dim=128, out_features=1):
        super().__init__()
        self.input_layer = nn.Sequential(
            nn.Linear(in_features, hidden_dim),
            nn.GELU()
        )
        self.res_block1 = ResidualBlock(hidden_dim, hidden_dim)
        # self.res_block2 = ResidualBlock(hidden_dim, hidden_dim)
        self.downsample = nn.Sequential(
            nn.Linear(hidden_dim, 64),
            nn.GELU(),
            nn.Linear(64, 32),
            nn.ReLU()
        )
        self.output_layer = nn.Linear(32, out_features)

    def forward(self, x):
        x = self.input_layer(x)
        x = self.res_block1(x)
        # x = self.res_block2(x)
        x = self.downsample(x)
        return self.output_layer(x)
```

```python
train_dataset = torch.utils.data.TensorDataset(torch.tensor(X_train, dtype=torch.float32), torch.tensor(y_train, dtype=torch.float32))
val_dataset = torch.utils.data.TensorDataset(torch.tensor(X_val, dtype=torch.float32), torch.tensor(y_val, dtype=torch.float32))
test_dataset = torch.utils.data.TensorDataset(torch.tensor(X_test, dtype=torch.float32), torch.tensor(y_test, dtype=torch.float32))

model = TabularNet(hidden_dim=512).to(device)

train_loader = torch.utils.data.DataLoader(dataset=train_dataset, batch_size=128, shuffle=True)
test_loader = torch.utils.data.DataLoader(dataset=test_dataset, batch_size=128)
val_loader = torch.utils.data.DataLoader(dataset=val_dataset, batch_size=128)

optimizer = torch.optim.AdamW(model.parameters(), lr=0.01)
criterion = nn.MSELoss()
epochs = 20
```
